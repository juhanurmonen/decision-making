```python
### Tuodaan kirjastot

# Tuodaan kirjastot

import pandas as pd

# Laskentaa varten

import numpy as np

#Kuvioita varten
import matplotlib
import matplotlib.pyplot as plt

#Ajanhallintaa
import datetime as dt
```

```python
remontti = pd.DataFrame({'Tehtävä' : ['Alku',
                                'Kaivuu',
                                'Perustus',
                                'Sadevesikaivot',
                                'Rungon_pystytys',
                                'Pohjakerroksen_putkityöt',
                                'Muuraus',
                                'Sähkötyöt',
                                'Pohjakerroksen_lattia',
                                'Putkityöt_runkoputket',
                                'Lämmitys',
                                'Vesikatto',
                                'Rappaus',
                                'Räystäät_rännit',
                                'Lattioiden_viimeistely',
                                'Pihan_tasoitus',
                                'Puutöiden_viimeistely',
                                'Keittiökalusteet',
                                'Putkitöiden_viimeistely',
                                'Käytävät_ja_puutarha',
                                'Maalaus',
                                'Lattioiden_lakkaus',
                                'Sähkötöiden_viimeistely',
                                'Loppu'],
                   'Kesto' : [0, 4, 2, 1, 4, 1, 6, 2, 2, 3, 4, 2, 10, 1, 3, 2, 3, 1, 2, 5, 3, 2, 1, 0]})
                   
```

```python
remontti
```

```python
## lisätään Te

remontti.loc[0, 'Te'] = remontti.loc[0, 'Kesto']
remontti.loc[1, 'Te'] = remontti.loc[0, 'Te'] + remontti.loc[1, 'Kesto']
remontti.loc[2, 'Te'] = remontti.loc[1, 'Te'] + remontti.loc[2, 'Kesto']
remontti.loc[3, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[3, 'Kesto']
remontti.loc[4, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[4, 'Kesto']
remontti.loc[5, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[5, 'Kesto']
remontti.loc[6, 'Te'] = remontti.loc[4, 'Te'] + remontti.loc[6, 'Kesto']
remontti.loc[7, 'Te'] = remontti.loc[4, 'Te'] + remontti.loc[7, 'Kesto']
remontti.loc[8, 'Te'] = remontti.loc[5, 'Te'] + remontti.loc[8, 'Kesto']
remontti.loc[9, 'Te'] = remontti.loc[5, 'Te'] + remontti.loc[9, 'Kesto']
remontti.loc[10, 'Te'] = max(remontti.loc[4, 'Te'], remontti.loc[8, 'Te'])+ remontti.loc[10, 'Kesto']
remontti.loc[11, 'Te'] = remontti.loc[6, 'Te'] + remontti.loc[11, 'Kesto']
remontti.loc[12, 'Te'] = max(remontti.loc[7, 'Te'], remontti.loc[10, 'Te'], remontti.loc[9, 'Te'])+ remontti.loc[12, 'Kesto']
remontti.loc[13, 'Te'] = remontti.loc[11, 'Te'] + remontti.loc[13, 'Kesto']
remontti.loc[14, 'Te'] = remontti.loc[12, 'Te'] + remontti.loc[14, 'Kesto']
remontti.loc[15, 'Te'] = max(remontti.loc[3, 'Te'], remontti.loc[13, 'Te'])+ remontti.loc[15, 'Kesto']
remontti.loc[16, 'Te'] = remontti.loc[14, 'Te'] + remontti.loc[16, 'Kesto']
remontti.loc[17, 'Te'] = remontti.loc[14, 'Te'] + remontti.loc[17, 'Kesto']
remontti.loc[18, 'Te'] = remontti.loc[14, 'Te'] + remontti.loc[18, 'Kesto']
remontti.loc[19, 'Te'] = remontti.loc[15, 'Te'] + remontti.loc[19, 'Kesto']
remontti.loc[20, 'Te'] = max(remontti.loc[17, 'Te'], remontti.loc[18, 'Te'])+ remontti.loc[20, 'Kesto']
remontti.loc[21, 'Te'] = max(remontti.loc[16, 'Te'], remontti.loc[20, 'Te'])+ remontti.loc[21, 'Kesto']
remontti.loc[22, 'Te'] = remontti.loc[20, 'Te'] + remontti.loc[22, 'Kesto']
remontti.loc[23, 'Te'] = max(remontti.loc[19, 'Te'], remontti.loc[21, 'Te'], remontti.loc[22, 'Te'])+ remontti.loc[23, 'Kesto']
```

```python
remontti
```

```python
## lisätään Tl

remontti.loc[23, 'Tl'] = remontti.loc[23, 'Te']
remontti.loc[22, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Kesto']
remontti.loc[21, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Kesto']
remontti.loc[20, 'Tl'] = min(remontti.loc[22, 'Tl'] - remontti.loc[22, 'Kesto'], remontti.loc[21, 'Tl'] - remontti.loc[21, 'Kesto'])
remontti.loc[19, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Kesto']
remontti.loc[18, 'Tl'] = remontti.loc[20, 'Tl'] - remontti.loc[20, 'Kesto']
remontti.loc[17, 'Tl'] = remontti.loc[20, 'Tl'] - remontti.loc[20, 'Kesto']
remontti.loc[16, 'Tl'] = remontti.loc[21, 'Tl'] - remontti.loc[21, 'Kesto']
remontti.loc[15, 'Tl'] = remontti.loc[19, 'Tl'] - remontti.loc[19, 'Kesto']
remontti.loc[14, 'Tl'] = min(remontti.loc[16, 'Tl'] - remontti.loc[16, 'Kesto'], remontti.loc[17, 'Tl'] - remontti.loc[17, 'Kesto'], remontti.loc[18, 'Tl'] - remontti.loc[18, 'Kesto'])
remontti.loc[13, 'Tl'] = remontti.loc[15, 'Tl'] - remontti.loc[15, 'Kesto']
remontti.loc[12, 'Tl'] = remontti.loc[14, 'Tl'] - remontti.loc[14, 'Kesto']
remontti.loc[11, 'Tl'] = remontti.loc[13, 'Tl'] - remontti.loc[13, 'Kesto']
remontti.loc[10, 'Tl'] = remontti.loc[12, 'Tl'] - remontti.loc[12, 'Kesto']
remontti.loc[9, 'Tl'] = remontti.loc[12, 'Tl'] - remontti.loc[12, 'Kesto']
remontti.loc[8, 'Tl'] = remontti.loc[10, 'Tl'] - remontti.loc[10, 'Kesto']
remontti.loc[7, 'Tl'] = remontti.loc[12, 'Tl'] - remontti.loc[12, 'Kesto']
remontti.loc[6, 'Tl'] = remontti.loc[11, 'Tl'] - remontti.loc[11, 'Kesto']
remontti.loc[5, 'Tl'] = min(remontti.loc[8, 'Tl'] - remontti.loc[8, 'Kesto'], remontti.loc[9, 'Tl'] - remontti.loc[9, 'Kesto'])
remontti.loc[4, 'Tl'] = min(remontti.loc[6, 'Tl'] - remontti.loc[6, 'Kesto'], remontti.loc[7, 'Tl'] - remontti.loc[7, 'Kesto'], remontti.loc[10, 'Tl'] - remontti.loc[10, 'Kesto'])
remontti.loc[3, 'Tl'] = remontti.loc[15, 'Tl'] - remontti.loc[15, 'Kesto']
remontti.loc[2, 'Tl'] = min(remontti.loc[3, 'Tl'] - remontti.loc[3, 'Kesto'], remontti.loc[4, 'Tl'] - remontti.loc[4, 'Kesto'], remontti.loc[5, 'Tl'] - remontti.loc[5, 'Kesto'])
remontti.loc[1, 'Tl'] = remontti.loc[2, 'Tl'] - remontti.loc[2, 'Kesto']
remontti.loc[0, 'Tl'] = remontti.loc[1, 'Tl'] - remontti.loc[1, 'Kesto']
```

```python
remontti['Pelivara'] = remontti['Tl'] - remontti['Te']

remontti
```

```python
plt.barh(y=remontti['Tehtävä'], width=remontti['Kesto'], left=(remontti['Te']-remontti['Kesto']))
plt.show()
```

```python

#Kuvan luonti ja akselien määrittely

fig, ax = plt.subplots()

plt.barh(y=remontti['Tehtävä'], width=remontti['Kesto'], left=(remontti['Te']-remontti['Kesto']))
plt.title('Remontin aikataulu', fontsize=15)

# Käännetään järjestys pystyakselilla

plt.gca().invert_yaxis()

# Määritellään vaaka-akselin 

xticks = np.arange(5, remontti['Tl'].max() + 2, 7)

ax.set_xticks(xticks)
# ax.set_xticklabels(xticklabels[::7])

# Ruudukko vaaka-akselille

ax.xaxis.grid(True, alpha=0.5)

# Lisätään tilaa oikeaan reunaan (x-akselin max-arvo + pieni lisä)

xlim_max = remontti['Tl'].max() + 2
ax.set_xlim(left=0, right=xlim_max)

plt.show()
```
